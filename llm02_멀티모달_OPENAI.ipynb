# 3.dall-e-3 이미지 생성

**dall-3**  
단순 편집 : v2  
생성 : v3

DALL-E-3 모델을 API로 호출하려면

```python
from openai import OpenAI
import requests

# dall-e-2 : 편집모델
# dall-e-3 : 생성모델


# 이미지 생성 요청
response = client.images.generate(
    prompt = 'a white siamese cat',
    #model = 'dall-e-3',
    model = 'gpt-image-2',
    size='1024x1024',
    quality='standard',
    n=1,
)

# 이미지 url 추출
image_url = response.data[0].url
print("image url: ", image_url)

# 이미지 다운로드 및 저장 (DALL-E 3기본 포맷인 png 확장자 권장)
image_data = requests.get(image_url).content
with open('output/genimg.png', 'wb') as fp :
    fp.write(image_data)

print('이미지가 output/genimg.png 경로에 성공적으로 저장')
```

# 명함 이미지에서 텍스트 읽어오기

```python
from openai import OpenAI
import base64   # 파이썬 표준 라이브러리

def encode_image(image_path) :
    with open(image_path, 'rb') as f:
        return base64.b64encode(f.read()).decode('utf-8')

image_base64 = encode_image('data/image/명함.png')  # 여기에 로컬 이미지 경로

data_url = f'data:image/png;base64,{image_base64}'

response = client.chat.completions.create(
    model='gpt-4o-mini',
    messages=[
        {'role':'user',
         'content' : [
             {'type':'text', 'text': '이미지의 글자 추출해 줄래'},
             {'type':'image_url', 'image_url': {'url': data_url }},
         ]}
    ]
)
print(response.choices[0].message.content)
```

# 7. 이미지에 대해 설명하기

```python
from openai import OpenAI
import requests

# 1. 위키미디어 이미지 다운로드(User_Agent 헤더 추가로 차단 회피)
url = 'https://www.yomidog.com/preSaleUpFile/230524_%EC%95%88%EC%82%B0%ED%82%B9%EC%B0%B0%EC%8A%A4_328.jpg'
headers = {'User-Agent' : 'Mozilla/5.0'}
img_response = requests.get(url, headers=headers)

# 2.이미지 바이너리를 Base64 문자열로 변환
base64_image = base64.b64encode(img_response.content).decode('utf-8')

# 3. Vision API 모양 (MiME 프리픽스 data:image/jpeg;base64, 포함
prompt = [
    {
        'role' : 'user',
        'content' : [
            {'type' : 'text', 'text':'견종 이름과 함께 이 이미지에 대해 설명해줘.'},
            {
                'type' : 'image_url',
                'image_url' : {
                    'url' : f'data:image/jpeg;base64, {base64_image}'
                }
            }
        ]
    }
]

response = client.chat.completions.create(
    model = 'gpt-4o-mini',
    messages=prompt
)


print(response.choices[0].message.content)
```
